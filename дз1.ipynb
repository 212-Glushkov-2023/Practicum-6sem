## Задача 2
хотим функцию,которая поворачивает матрицу на 90градусов. zip

```python
def rot_mat(m,d='right'):
    if d =='right':
        return [list(row) for row in zip(*m[::-1])]
    elif d=='left':
        return [list(row) for row in list(zip(*m))[::-1]]
    return m
ex=[[1,2,3],[4,5,6],[7,8,9]]
print("Поворот влево:",rot_mat(ex,'left'))
print("Поворот вправо:",rot_mat(ex,'right'))
```

## Задача 4

система управления библиотекой 

```python
class Book:
    def __init__(self, title, author, year, isbn, copies):
        self.title = title
        self.author = author
        self.year = year
        self.isbn = isbn
        self.copies = copies
    def info(self):
        return f"{self.title} by {self.author} ({self.year}) - {self.copies} copies left"
    def change_copies(self, delta):
        self.copies += delta
class User:
    def __init__(self, name, card_id):
        self.name = name
        self.card_id = card_id
        self.borrowed = []
    def borrow_book(self, book):
        self.borrowed.append(book)
    def return_book(self, book):
        self.borrowed.remove(book)






class Library:
    def __init__(self):
        self.books = []
        self.users = []
        self.history = []
    def add_book(self, book):
        self.books.append(book)
    def remove_book(self, isbn):
        self.books = [b for b in self.books if b.isbn != isbn]
    def add_user(self, user):
        self.users.append(user)
    def remove_user(self, card_id):
        self.users = [u for u in self.users if u.card_id != card_id]
    def find_book(self, title=None, author=None):
        return [b for b in self.books if (title is None or b.title == title) and (author is None or b.author == author)]
    def find_user(self, name):
        return [u for u in self.users if u.name == name]
    def issue_book(self, user, book):
        if book.copies > 0:
            book.change_copies(-1)
            user.borrow_book(book)
            self.history.append(f"{user.name} borrowed '{book.title}'")
            return True
        return False
    def return_book(self, user, book):
        if book in user.borrowed:
            book.change_copies(1)
            user.return_book(book)
            self.history.append(f"{user.name} returned '{book.title}'")
            return True
        return False
    def show_history(self):
        for event in self.history:
            print(event)
lib = Library()
book1 = Book("The Hobbit", "J.R.R. Tolkien", 1937, "9780547928227", 5)
book2 = Book("Dune", "Frank Herbert", 1965, "9780441013593", 3)
book3 = Book("Neuromancer", "William Gibson", 1984, "9780441569595", 2)
user1 = User("Arthur Morgan", "L001")
user2 = User("Miss Grimsho", "L002")
user3 = User("Dutch Van Der Linde", "L003")
lib.add_book(book1)
lib.add_book(book2)
lib.add_book(book3)
lib.add_user(user1)
lib.add_user(user2)
lib.add_user(user3)
lib.issue_book(user1, book1)  
lib.issue_book(user2, book2)   
lib.issue_book(user3, book3)   
lib.return_book(user1, book1) 
lib.issue_book(user2, book3)   
lib.return_book(user2, book2)  
print("\n=== История операций ===")
lib.show_history()
print("\n=== Остатки книг ===")
for b in [book1, book2, book3]:
    print(b.info())
```

## Задача 5

почему после выполнения кода ниже переменная b изменилась:


a = [1, 2, 3]
b = a
a[0] = 4
print(b)
     


Концепция:

Переменные в питоне хранят ссылки на объекты, а не сами значения. если мы присваиваем b = a, то при этом обе переменные начинают указывать на один и тот же список в памяти. Изменение списка - когда мы прописываем a[0] меняет общий объект, поэтому b тоже «видит» это изменение.

## Задача 6



```python
sum1=0
for i in range(1,10001):
  sum1+=1/(i**2)
print(sum1)
sum2=0
for i in range(10000,0,-1):
  sum2+=1/(i**2)
print(sum2)
```

Концепция: Значения sum1 и sum2 различаются из-за потери точности при сложении чисел с плавающей точкой. В обычном порядке сложения маленькие слагаемые не заметны в большой накопленной сумме. В обратном порядке сначала складываются малые числа, поэтому они учитываются полностью. Вывод: обратный порядок суммирования точнее.
